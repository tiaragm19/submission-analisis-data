## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana tren revenue bulanan (total price pesanan berstatus delivered) pada 5 kategori produk dengan revenue tertinggi selama Januari 2017 - Agustus 2018, agar tim bisnis dapat menentukan kategori yang diprioritaskan untuk promosi?
- **Pertanyaan 2:** Pada state pelanggan mana persentase pesanan yang tiba melewati estimasi pengiriman paling tinggi selama Januari 2017 - Agustus 2018, dan berapa selisih rata-rata review score antara pesanan terlambat dan tepat waktu, agar tim logistik dapat menentukan wilayah yang diperbaiki lebih dulu?
- **Pertanyaan 3:** Berapa proporsi pelanggan dan kontribusi revenue dari tiap segmen RFM (Recency, Frequency, Monetary) berdasarkan pesanan berstatus delivered selama Januari 2017 - Agustus 2018, dan segmen mana yang menyumbang revenue terbesar, agar tim marketing dapat menentukan segmen yang diprioritaskan untuk kampanye retensi?

### Cleaning Data

#### Fixing tipe data kolom tanggal pada tabel orders

```python
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]

orders_clean = orders_df.copy()
for col in date_cols:
    orders_clean[col] = pd.to_datetime(orders_clean[col])

orders_clean.info()
```

#### Fixing pesanan tidak relevan, tanggal tiba kosong, dan periode di luar analisis

```python
print("Jumlah baris awal:", len(orders_clean))

# Hanya pesanan yang sudah sampai ke pelanggan
orders_clean = orders_clean[orders_clean.order_status == "delivered"]
print("Setelah filter delivered:", len(orders_clean))

# Buang pesanan delivered yang tidak punya tanggal tiba
orders_clean = orders_clean.dropna(subset=["order_delivered_customer_date"])
print("Setelah buang tanggal tiba kosong:", len(orders_clean))

# Batasi periode analisis: Januari 2017 - Agustus 2018
orders_clean = orders_clean[
    (orders_clean.order_purchase_timestamp >= "2017-01-01")
    & (orders_clean.order_purchase_timestamp < "2018-09-01")
]
print("Setelah batasi periode:", len(orders_clean))

print("\nMissing value pada kolom kunci:")
print(orders_clean[["order_purchase_timestamp", "order_delivered_customer_date"]].isna().sum())
```

#### Menambahkan metrik keterlambatan pengiriman

```python
orders_clean["delay_days"] = (
    orders_clean.order_delivered_customer_date - orders_clean.order_estimated_delivery_date
).dt.total_seconds() / 86400

orders_clean["is_late"] = orders_clean.delay_days > 0

print("Persentase pesanan terlambat:", round(orders_clean.is_late.mean() * 100, 2), "%")
orders_clean[["order_id", "order_delivered_customer_date", "order_estimated_delivery_date", "delay_days", "is_late"]].head()
```

#### Fixing review ganda pada tabel order_reviews

```python
reviews_df["review_answer_timestamp"] = pd.to_datetime(reviews_df.review_answer_timestamp)

reviews_clean = (
    reviews_df.sort_values("review_answer_timestamp")
    .drop_duplicates(subset="order_id", keep="last")[["order_id", "review_score"]]
)

print("Baris sebelum:", len(reviews_df))
print("Baris sesudah:", len(reviews_clean))
print("order_id ganda setelah cleaning:", reviews_clean.order_id.duplicated().sum())
```

#### Fixing kategori kosong dan terjemahan yang hilang pada tabel products

```python
# Terjemahan manual untuk 2 kategori yang tidak ada di tabel translation
manual_translation = pd.DataFrame({
    "product_category_name": ["pc_gamer", "portateis_cozinha_e_preparadores_de_alimentos"],
    "product_category_name_english": ["pc_gamer", "portable_kitchen_food_preparers"],
})
category_full = pd.concat([category_df, manual_translation], ignore_index=True)

products_clean = products_df.merge(category_full, on="product_category_name", how="left")
products_clean["product_category_name_english"] = products_clean.product_category_name_english.fillna("unknown")
products_clean = products_clean[["product_id", "product_category_name_english"]]

print("Baris products sebelum:", len(products_df))
print("Baris products sesudah:", len(products_clean))
print("Kategori kosong (NaN):", products_clean.product_category_name_english.isna().sum())
print("Produk berkategori unknown:", (products_clean.product_category_name_english == "unknown").sum())
```

```python
main_df = (
    order_items_df
    .merge(orders_clean[["order_id", "customer_id", "order_purchase_timestamp",
                         "order_delivered_customer_date", "order_estimated_delivery_date",
                         "delay_days", "is_late"]], on="order_id", how="inner")
    .merge(products_clean, on="product_id", how="left")
    .merge(customers_df[["customer_id", "customer_unique_id", "customer_city", "customer_state"]],
           on="customer_id", how="left")
    .merge(reviews_clean, on="order_id", how="left")
)

# Validasi: tidak ada item yang terduplikasi akibat merge
print("Item duplikat:", main_df.duplicated(subset=["order_id", "order_item_id"]).sum())
print("Baris main_df:", len(main_df))
print("Pesanan unik:", main_df.order_id.nunique())
print("Total revenue (price):", round(main_df.price.sum(), 2))
print("\nMissing value:")
print(main_df.isna().sum()[main_df.isna().sum() > 0])
main_df.head()
```

## Exploratory Data Analysis (EDA)

### Explore revenue per kategori produk

```python
main_df["order_month"] = main_df.order_purchase_timestamp.dt.to_period("M")

category_summary = (
    main_df.groupby("product_category_name_english")
    .agg(
        revenue=("price", "sum"),
        total_orders=("order_id", "nunique"),
        total_items=("order_item_id", "count"),
        avg_price=("price", "mean"),
    )
    .sort_values("revenue", ascending=False)
)
category_summary["revenue_share_pct"] = category_summary.revenue / category_summary.revenue.sum() * 100

category_summary.head(10).round(2)
```

### Explore tren revenue bulanan 5 kategori teratas

```python
top5 = category_summary.head(5).index

monthly_top5 = (
    main_df[main_df.product_category_name_english.isin(top5)]
    .groupby(["order_month", "product_category_name_english"])
    .price.sum()
    .unstack()
    .fillna(0)
)

monthly_top5.round(0)
```

```python
print("Kontribusi 5 kategori teratas:", round(category_summary.head(5).revenue_share_pct.sum(), 2), "%")
print("Jumlah kategori:", len(category_summary))
category_summary.loc["unknown"].round(2)
```

### Explore keterlambatan pengiriman per state

```python
order_df = main_df.drop_duplicates(subset="order_id")[
    ["order_id", "customer_state", "is_late", "delay_days", "review_score"]
].copy()

print("Jumlah pesanan:", len(order_df))
print("Persentase terlambat keseluruhan:", round(order_df.is_late.mean() * 100, 2), "%")
```

```python
# Persentase terlambat per state
state_delay = (
    order_df.groupby("customer_state")
    .agg(
        total_orders=("order_id", "nunique"),
        late_orders=("is_late", "sum"),
        late_pct=("is_late", "mean"),
        avg_review=("review_score", "mean"),
    )
    .sort_values("late_pct", ascending=False)
)
state_delay["late_pct"] = state_delay.late_pct * 100

state_delay.round(2).head(10)
```

### Explore review score pesanan terlambat vs tepat waktu

```python
review_compare = order_df.groupby("is_late").review_score.agg(["mean", "count"]).round(3)
print(review_compare)
print("Selisih rata-rata:", round(review_compare.loc[False, "mean"] - review_compare.loc[True, "mean"], 3))
```

**Binning keterlambatan:** nilai keterlambatan (delay_days) dikelompokkan ke dalam interval (tepat waktu atau lebih awal, 1-3 hari, 4-7 hari, lebih dari 7 hari) dengan pd.cut. Tujuannya melihat apakah dampak keterlambatan terhadap review score meningkat seiring lamanya keterlambatan, yang tidak terlihat bila hanya membandingkan terlambat dan tepat waktu.

```python
# Mengetahui seberapa parah keterlambatan memengaruhi review
order_df["delay_group"] = pd.cut(
    order_df.delay_days,
    bins=[-np.inf, 0, 3, 7, np.inf],
    labels=["Tepat waktu / lebih awal", "Terlambat 1-3 hari", "Terlambat 4-7 hari", "Terlambat >7 hari"],
)

order_df.groupby("delay_group", observed=True).agg(
    total_orders=("order_id", "count"),
    avg_review=("review_score", "mean"),
).round(2)
```

**Insight:**
- Secara keseluruhan, 8,13% dari 96.203 pesanan tiba melewati estimasi pengiriman (sekitar 7.822 pesanan).
- Persentase keterlambatan tertinggi ada di AL (23,99% dari 396 pesanan), MA (19,64% dari 713 pesanan), PI (16,00%), CE (15,40%), dan SE (15,36%). Seluruh state di sepuluh besar memiliki persentase di atas 12%, jauh di atas rata-rata nasional 8,13%.
- State dengan jumlah pesanan kecil (misalnya RR 40 pesanan, TO 274 pesanan) ikut masuk sepuluh besar, sehingga persentasenya kurang stabil dan perlu ditafsirkan hati-hati. Di antara state berukuran besar, RJ menonjol dengan 13,52% dari 12.310 pesanan (1.664 pesanan terlambat), dan BA 14,05% dari 3.253 pesanan.
- Pesanan yang tiba tepat waktu atau lebih awal memiliki rata-rata review score 4,30 (87.902 pesanan), sedangkan pesanan terlambat hanya 2,57 (7.658 pesanan yang memiliki review). Selisihnya 1,73 poin.
- Dampak ke review makin besar seiring lamanya keterlambatan: rata-rata review 4,29 untuk tepat waktu, 3,77 untuk terlambat 1-3 hari, 2,32 untuk 4-7 hari, dan 1,73 untuk lebih dari 7 hari. Penurunan paling tajam terjadi saat keterlambatan melewati 3 hari.

## Analisis Lanjutan (Opsional)

### RFM Analysis

**Tujuan RFM Analysis:** mengelompokkan pelanggan berdasarkan perilaku belanjanya agar tim marketing dapat menentukan segmen yang diprioritaskan untuk kampanye retensi (Pertanyaan 3). RFM memakai tiga ukuran: Recency (berapa hari sejak pembelian terakhir), Frequency (jumlah pesanan), dan Monetary (total belanja). Teknik ini dipilih karena dataset berisi riwayat transaksi per pelanggan, dan dilakukan tanpa algoritma machine learning.

```python
# Menghitung R, F, M per pelanggan

# Identitas pelanggan = customer_unique_id (bukan customer_id)
rfm_df = (
    main_df.groupby("customer_unique_id")
    .agg(
        last_purchase=("order_purchase_timestamp", "max"),
        frequency=("order_id", "nunique"),
        monetary=("price", "sum"),
    )
    .reset_index()
)

# Tanggal acuan: sehari setelah transaksi terakhir pada data
ref_date = main_df.order_purchase_timestamp.max().normalize() + pd.Timedelta(days=1)
rfm_df["recency"] = (ref_date - rfm_df.last_purchase).dt.days

print("Tanggal acuan:", ref_date.date())
print("Jumlah pelanggan:", len(rfm_df))
print(rfm_df.frequency.value_counts().sort_index().head(5))
rfm_df[["recency", "frequency", "monetary"]].describe().round(1)
```

**Aturan segmentasi RFM (manual grouping, tanpa machine learning):**
- Pelanggan diidentifikasi dengan customer_unique_id, dengan tanggal acuan 30 Agustus 2018 (sehari setelah transaksi terakhir pada data).
- Recency dan Monetary dibagi menjadi 3 kelompok berdasarkan kuantil (skor 1-3). Frequency memakai aturan manual karena sekitar 97% pelanggan hanya membeli sekali.
- Loyal Customers: membeli 2 kali atau lebih.
- Pelanggan yang membeli sekali dibagi berdasarkan recency dan monetary: New High Spenders (recency terbaru dan monetary tinggi), Recent Customers (recency terbaru, monetary lebih rendah), At-Risk High Spenders (sudah lama tidak bertransaksi dan monetary tinggi), serta Lost / Low Value (sudah lama tidak bertransaksi dan monetary lebih rendah).

```python
# Beri skor dan tentukan segmen (manual grouping)

# Recency dan Monetary dibagi 3 kelompok (kuantil), Frequency memakai aturan manual
rfm_df["r_score"] = pd.qcut(rfm_df.recency, 3, labels=[3, 2, 1]).astype(int)   # makin baru makin tinggi
rfm_df["m_score"] = pd.qcut(rfm_df.monetary, 3, labels=[1, 2, 3]).astype(int)

def tentukan_segmen(row):
    if row.frequency >= 2:
        return "Loyal Customers"
    if row.r_score == 3:
        return "New High Spenders" if row.m_score == 3 else "Recent Customers"
    return "At-Risk High Spenders" if row.m_score == 3 else "Lost / Low Value"

rfm_df["segment"] = rfm_df.apply(tentukan_segmen, axis=1)
```

```python
# Ringkasan per Segmen
segment_summary = (
    rfm_df.groupby("segment")
    .agg(
        customers=("customer_unique_id", "count"),
        revenue=("monetary", "sum"),
        avg_recency=("recency", "mean"),
        avg_monetary=("monetary", "mean"),
    )
)
segment_summary["customer_pct"] = segment_summary.customers / segment_summary.customers.sum() * 100
segment_summary["revenue_pct"] = segment_summary.revenue / segment_summary.revenue.sum() * 100

segment_summary.sort_values("revenue_pct", ascending=False).round(2)
```

```python
seg_plot = segment_summary.sort_values("revenue_pct", ascending=False).reset_index()
seg_long = seg_plot.melt(id_vars="segment", value_vars=["customer_pct", "revenue_pct"],
                         var_name="ukuran", value_name="persen")
seg_long["ukuran"] = seg_long.ukuran.map({"customer_pct": "% pelanggan", "revenue_pct": "% revenue"})

top_seg = seg_plot.iloc[0]
fig, ax = plt.subplots(figsize=(10, 5.5))
sns.barplot(data=seg_long, x="persen", y="segment", hue="ukuran",
            palette={"% pelanggan": "#c7c7c7", "% revenue": "#1f77b4"}, ax=ax)
for p in ax.patches:
    w = p.get_width()
    if w > 0:
        ax.text(w + 0.5, p.get_y() + p.get_height() / 2, f"{w:.1f}%", va="center", fontsize=9)
ax.set_title(f"{top_seg.segment}: {top_seg.customer_pct:.0f}% pelanggan, {top_seg.revenue_pct:.0f}% revenue",
             fontsize=13, loc="left")
ax.set_xlabel("Persentase dari total (pelanggan atau revenue)")
ax.set_ylabel("")
ax.set_xlim(0, seg_long.persen.max() * 1.15)
ax.legend(title="", loc="lower right")
plt.tight_layout()
plt.show()
```

**Insight:**
- Retensi pelanggan sangat rendah: dari 93.096 pelanggan, 90.307 (sekitar 97%) hanya membeli satu kali. Segmen Loyal Customers hanya 2.789 pelanggan (3,00%) dan menyumbang 5,50% revenue.
- At-Risk High Spenders adalah segmen paling bernilai: 19.310 pelanggan (20,74%) menyumbang 43,30% revenue dengan rata-rata belanja 295,53, tetapi rata-rata transaksi terakhirnya sudah 316 hari lalu.
- New High Spenders (9.780 pelanggan atau 10,51%) menyumbang 22,52% revenue dengan rata-rata belanja 303,54 dan transaksi terakhir rata-rata 77 hari lalu. Gabungan kedua segmen high spender ini hanya 31,25% pelanggan tetapi menyumbang 65,82% revenue.
- Lost / Low Value adalah segmen terbesar (40.830 pelanggan atau 43,86%) namun hanya menyumbang 18,98% revenue, dengan rata-rata belanja 61,27 dan transaksi terakhir rata-rata 318 hari lalu.
- Recent Customers (20.387 pelanggan atau 21,90%) baru bertransaksi sekitar 75 hari lalu dengan rata-rata belanja 62,65, sehingga masih berpeluang dikonversi menjadi pelanggan berulang.
